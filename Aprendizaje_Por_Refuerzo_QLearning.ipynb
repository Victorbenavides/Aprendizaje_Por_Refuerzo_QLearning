```python
#Se aplican las librerias a usar
import numpy as np#crear vectores y matrices grandes multidimensionales

class Env: #Se crea una clase llamada Env
    def __init__(self):#init: inicializar los atributos del objeto,init:cargar los atributos,self:hace referencia al nombre del objeto.
      #tamaño del mapa
        self.height = 11#altura
        self.width = 11#anchura
        #posicion del agente dentro del mapa
        self.posX = 0#se posiciona en 0
        self.posY = 0#se posiciona en 0
        #posicion de la meta dentro del mapa
        self.endX = self.width-1#se posiciona la meta final en -1 para anchura
        self.endY = self.height-1#se posiciona la meta final en -1 para altura
        #4 posibles acciones
        self.actions = [0, 1, 2, 3]#Las cuatro posibles posiciones
        #el numero de posibles estados es el tamaño del mapa
        self.stateCount = self.height*self.width
        self.actionCount = len(self.actions)

    def reset(self): #Se define la funcion rese t
        self.posX = 0 #Posicion X a 0
        self.posY = 0 #Posicion Y a 0
        self.done = False
        return 0, 0, False

    # Acciones
    def step(self, action):
        if action == 0: # moverse a la izquierda
            self.posX = self.posX-1 if self.posX > 0 else self.posX
        if action == 1: # moverse a la derecha
            self.posX = self.posX+1 if self.posX < self.width - 1 else self.posX
        if action == 2: # moverse arriba
            self.posY = self.posY-1 if self.posY > 0 else self.posY
        if action == 3: # moverse abajo
            self.posY = self.posY+1 if self.posY < self.height - 1 else self.posY

        done = self.posX == self.endX and self.posY == self.endY
        # mapping (x,y) position to number between 0 and 11x11-1=120
        nextState = self.width * self.posY + self.posX
        reward = 1 if done else 0
        return nextState, reward, done

    # Retorna una accion aleatoria
    def randomAction(self):
        return np.random.choice(self.actions)

    # mostramos medio ambiente
    def render(self):
        for i in range(self.height):#ciclo anidado para las casillas de alto y ancho
            for j in range(self.width):
                if self.posY == i and self.posX == j:
                    print("O", end='')#imprimimos 0 para la posicion del agnete
                elif self.endY == i and self.endX == j:
                    print("F", end='')#imprimimos estado final
                else:#si no imprimimos puntos que representan casillas vacias
                    print(".", end='')
            print("")
```

```python
#se aplican las librerias a usar
import numpy as np
import time
import os

#llama a constructor de la clase Env
env = Env()

# QTable : contiene los valores para cada par (state,action)
# inicializacion aleatoria, distribucion uniforme entre cero y uno
qtable = np.random.rand(env.stateCount, env.actionCount).tolist()

# parametros
#cuantos juegos va a realizar para entrenarse
epochs = 20
#factor de descuento
gamma = 0.1
#seleccionar entre exploracion y explotacion
epsilon = 0.08
#decaimiento de epsilon en cada epoca (disminuir exploracion conforme se)
decay = 0.1

# ciclo de entrenamiento
for i in range(epochs):
    state, reward, done = env.reset()
    steps = 0

    while not done:
        os.system('clear')
        print("epoch #", i+1, "/", epochs)
        env.render()
        time.sleep(0.05)

        # cuenta los pasos hasta terminar el juego
        steps += 1

        # actuar al azar a veces para permitir la exploración
        if np.random.uniform() < epsilon:
            action = env.randomAction()
        # si no, seleccione la acción máxima en Qtable (actúe codicioso)

        else:
            action = qtable[state].index(max(qtable[state]))

        # toma acciones
        next_state, reward, done = env.step(action)

        # actualizar el valor de qtable con la ecuación de Bellman
        qtable[state][action] = reward + gamma * max(qtable[next_state])

        # actualiza el estado
        state = next_state
    # Cuanto más aprendemos, menos tomamos acciones aleatorias
    epsilon -= decay*epsilon

    print("\nDone in", steps, "steps".format(steps)) #Se muestran los pasos
    time.sleep(0.8) #Se aplica un tiempo
```
